```python
#Create variables for student details and display them.

student_name = "Prerna"
student_age = 19
student_grade = "A"
student_id = 867

print("Student Name:", student_name)
print("Student Age:", student_age)
print("Student Grade:", student_grade)
print("Student ID:", student_id)
```

```python
#Find the datatype of different variables using type().
student_name = "Prerna"
student_age = 19
student_percentage = 85.5
is_student = True

print(type(student_name))
print(type(student_age))
print(type(student_percentage))
print(type(is_student))
```

```python
#Swap two numbers without using a temporary variable.

a = 10
b = 20
a, b = b, a
print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
#Demonstrate multiple assignment and unpacking.

a = b = c = 10
print("a =", a)
print("b =", b)
print("c =", c)
name, age, branch = "Prerna", 19, "CSE"

print("Name:", name)
print("Age:", age)
print("Branch:", branch)
```

```python
#Compare id() values before and after reassignment.

a = 10
print("ID before reassignment:", id(a))
a = 20
print("ID after reassignment:", id(a))
```

```python
#Write a program to show mutable vs immutable behavior.

a = [1, 2, 3]
print("Before:", a)
print("ID:", id(a))

a.append(4)
print("After:", a)
print("ID:", id(a))

b = "Hello"
print("Before:", b)
print("ID:", id(b))

b = b + " World"
print("After:", b)
print("ID:", id(b))
```

```python
#Create a global variable and modify it inside a function.

x = 10

def modify():
    global x
    x = 20

print("Before:", x)
modify()
print("After:", x)
```

```python
#Use del to remove a variable and observe the result.

x = 10

print("Before deleting:", x)

del x

try:
    print("After deleting:", x)
except NameError:
    print("Variable x has been deleted")
```
